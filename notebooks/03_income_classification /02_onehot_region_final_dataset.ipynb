# Import Libraries

```python
import pandas as pd
```

# Load dataset

```python
dataset = pd.read_csv('datset_preprocessed_after_cleaning.csv')
dataset.head()
```

# Encode region

```python
df_final = dataset
```

```python
print("\nRegion distribution before one-hot:")
print(df_final["region"].value_counts(dropna=False))

df_final = pd.get_dummies(
    df_final,
    columns=["region"],
    prefix="region",
    dtype=int
)

region_cols = [c for c in df_final.columns if c.startswith("region_")]
print("\nRegion columns created:", region_cols)
```

# Quick Check :3

```python
print("\nFinal columns:")
print(df_final.columns.tolist())

print("\nMissing values in final dataset:")
print(df_final.isnull().sum().sort_values(ascending=False).head(10))
```

# Save the final dataset

```python
df_final.to_csv("final_dataset_preprocessed.csv",index=False)
```
